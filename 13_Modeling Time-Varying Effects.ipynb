# Event Studies with Time-Varying Effects
Not all treatments have a constant effect across time. Event studies can be extended to capture period-by-period effects, revealing richer dynamics.

## Why It Matters
- Novelty effects: Spike in engagement that fades quickly.
- Ramp-up effects: Gradual increase, such as ad campaigns or feature launches.
- Business insight: Understanding whether an impact is short-lived or sustained is crucial for strategy.


## How It Works



1. Replace the single post-event indicator with multiple dummy variables:
- post0 = 1 if outcome is measured the day of the event.
- post1 = 1 if the next period, etc.
- Each dummy allows its own estimated effect.
  


2. Fit an OLS model with the treated group as a function of:
- A comparison group.
- The set of post-period indicators.

3. Interpret coefficients:
- Large post0, small later coefficients → short-term spike.
- Coefficients that grow over time → ramp-up or compounding effect.



## Takeaway

Time-varying event studies provide a timeline of treatment effects, not just a single number. Plotting coefficients across periods helps communicate whether an effect was immediate, delayed, or sustained.

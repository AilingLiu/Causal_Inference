# Placebo Testing in Causal Inference

Placebo testing helps to validate whether a model is detecting real treatment effects or just noise/confounding:

- Pretend the treatment happened earlier, or assign it to a random group.



- If the model finds large/significant effects under these fake scenarios, it suggests bias or misspecification.
- Running multiple placebo tests (different fake dates or random assignments) creates a distribution of estimates.



- If the actual treatment effect stands out compared to these placebos, confidence in the causal result increases.

Placebo testing is a versatile validation tool across many causal models.

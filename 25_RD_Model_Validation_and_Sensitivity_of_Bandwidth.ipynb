# Assumptions in Regression Discontinuity
For regression discontinuity (RD) to yield credible causal effects, two key assumptions must hold:

- Local randomization: units just above and below the cutoff are comparable, as if randomly assigned to treatment and control.
- Continuity of potential outcomes: in the absence of treatment, the outcome would evolve smoothly across the cutoff, with no sharp break.

## Local Randomization: Covariate Balance Checks
- Restrict the sample to units within a chosen bandwidth around the cutoff.
- Use t-tests to compare mean values of covariates (e.g., age, income, prior engagement).
    - No significant differences → supports local randomization.
    - Significant differences → red flag; investigate whether groups differ systematically.

## Continuity: Placebo Cutoff Tests
- Select a set of fake thresholds where no treatment occurred.
- Re-estimate the RD model at each placebo cutoff.
- Plot estimated treatment effects across these placebo points.
    - If continuity holds: placebo effects should be close to zero.
    - If large jumps appear: could indicate natural breaks in the outcome not caused by treatment.

Together, these diagnostics increase confidence that the estimated jump at the real cutoff reflects a true causal effect rather than bias or noise.

## Bandwidth Choice in Regression Discontinuity
A central modeling decision in regression discontinuity (RD) is how much data to include around the cutoff, known as the bandwidth. The bandwidth determines which observations are considered “close enough” to the threshold to be comparable.

## The Bias–Variance Tradeoff
- Narrow bandwidths: focus only on units right at the cutoff. This strengthens the local randomization assumption but leaves fewer observations, which can make estimates noisy.
- Wider bandwidths: include more observations, improving precision but risking bias if units farther away are systematically different.
This tradeoff mirrors hyperparameter choices in other models—like the pre-period window in DiD or the donor pool in synthetic controls.

## Sensitivity Testing
There is no single correct bandwidth. Instead, analysts re-estimate the RD model with multiple windows (e.g., ±5, ±10, ±15 units) and compare results.

- Stable estimates across reasonable bandwidths → increases confidence in robustness.
- Large shifts with wider windows → may indicate that units farther from the cutoff are not truly comparable.
- There is a body of research related to bandwidth selection. For example:
    - [Imbens, G., & Kalyanaraman, K. (2012). Optimal bandwidth choice for the regression discontinuity estimator. The Review of economic studies, 79(3), 933-959](https://www.nber.org/system/files/working_papers/w14726/w14726.pdf).
    - [Calonico, S., Cattaneo, M. D., & Farrell, M. H. (2020). Optimal bandwidth choice for robust bias-corrected inference in regression discontinuity designs. The Econometrics Journal, 23(2), 192-210.](https://arxiv.org/pdf/1809.00236)

## Manipulation at the Cutoff
In practice, individuals may try to game the threshold:

- Applicants improving credit scores just above 700 to qualify for loans.
- Students receiving grade bumps to cross a pass/fail boundary.
- Taxpayers reporting death dates to avoid higher estate taxes.
[Kopczuk, W., & Slemrod, J. (2003). Dying to save taxes: Evidence from estate-tax returns on the death elasticity. Review of Economics and Statistics, 85(2), 256-265.](https://www.nber.org/system/files/working_papers/w8158/w8158.pdf)

These behaviors can **create clumping near the cutoff**, undermining local randomization.

## Donut Hole Designs
If manipulation is suspected, one solution is a donut hole design: remove units very close to the cutoff and re-estimate the model. This avoids the manipulated zone, restoring credibility—though at the cost of fewer observations.

In short, bandwidth choice is both a modeling decision and a diagnostic tool. Robust results across plausible bandwidths support causal claims, while instability or manipulation near the cutoff calls for caution.

## Resources
The donut hole design in regression discontinuity excludes observations very close to the cutoff to address potential manipulation or sorting.

- Original application: Barreca et al. (2011) introduced the donut RD design to reduce bias when individuals could manipulate their status just around the threshold. [Saving Babies? The Impact of Public Health Insurance on Infant Mortality](https://www.nber.org/papers/w17658)
- Recent analysis: Noack & Rothe (2023) formally analyzed the method, showing that while the donut design helps guard against sorting artifacts, it can increase bias and variance, leading to wider confidence intervals. [The Properties of Donut Designs in Regression Discontinuity](https://arxiv.org/pdf/2308.14464)

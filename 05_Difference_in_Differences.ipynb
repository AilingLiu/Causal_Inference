# Difference in Differences

## Introduction to Difference-in-Differences (DiD)

## Why DiD?
Interrupted Time Series (ITS) works when everyone is treated at once, but it can fail when time trends are noisy or other events overlap with the treatment.

## What DiD Does
- Compares outcomes over time between a treated group and an untreated group.
- Helps isolate the true causal effect of the treatment.

## Key Analyst Habits
- Test assumptions using pre-event data.
- Use visual and statistical diagnostics to validate trends.
- Run placebo tests to detect false positives.

## Why ITS Can Fall Short
Interrupted Time Series (ITS) can estimate treatment effects when everyone is treated at once.

## But it struggles when:

- Outcomes follow hard-to-predict nonlinear trends (holidays, seasons, school cycles).
- Other events overlap with the treatment (ads, competitor promotions).

## The Problem
- Assuming a straight-line trend risks misattributing natural fluctuations.
- Ignoring simultaneous shocks can over- or underestimate the true effect.

## The Solution
- Introduce a comparison group that:
- Shares seasonal patterns and outside influences.
- Was not exposed to the treatment.
- This provides a more reliable estimate of causal impact.

When trends are messy or outside shocks occur, comparison groups are essential—motivating the Difference-in-Differences approach introduced next.

The difference-in-differences method is essential for controlling external factors and isolating treatment effects.

## Introduction to Difference-in-Differences (DiD)

DiD isolates causal effects by leveraging comparison groups over time.

## Core Idea
Compare changes over time in an outcome for:

- A treated group
- An untreated (comparison) group
- If the treated group changes more than the comparison group, that extra difference is attributed to the treatment.

## How It Works
- **Pre-treatment**: Assume both groups would have followed similar trends.
- **Post-treatment**: Measure how the untreated group changes.
- **Counterfactual**: Use that change to estimate what would have happened to the treated group without treatment (Y(0)).
- **Causal Effect**: The gap between the treated group’s actual outcome and its counterfactual gives the treatment effect.

In other words, the following equation represents the DiD Treatment Effect:

$DID Treatment Effect = (Y_{treated_{post}}-Y_{treated_{pre}})-(Y_{untreated_{post}}-Y_{untreated_{pre}})$



## Model Set Up
we estimate causal erects using a regression or the form.

- $Treatment_i$;: 1 if unit is in the treated group, 0 otherwise
- $Post_t$: 1 if time t is after the treatment date, 0 otherwise
- $Treatment_i \times Post_t$: interaction term equal to 1 only for treated units in the post-treatment period

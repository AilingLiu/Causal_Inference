## Building Trustworthy Causal Inference Without Experiments
In practice, we often can’t run clean randomized experiments. But by applying rigorous principles, we can still make credible causal claims and persuade stakeholders to trust our findings.

## Three Core Principles
- Choose appropriate models for the data and context.
- Think critically about assumptions and test them whenever possible.
- Demonstrate robustness through validation and sensitivity checks.

## Critical Assumptions by Model
Each causal method has a key assumption that underpins its validity:

- Difference-in-Differences (DiD): parallel trends
- Event studies: stable relative relationships
- Regression discontinuity (RD): local randomization and continuity
- Synthetic control: comparability of the donor pool

## Tools for Validation
Analysts can build confidence by checking assumptions with diagnostics such as:

- Pre-period indicators to detect anticipatory effects
- Covariate balance checks near the cutoff (RD)
- Placebo tests at fake treatment dates or cutoffs
- Visual inspection of trends and overlaps in time series data

## Robustness and Sensitivity
Stakeholders want to know if results are fragile. Useful strategies include:

- Trying multiple bandwidths in RD.
- Varying pre-period lengths in DiD.
- Adjusting donor pools in synthetic controls, such as leaving out each comparison unit one by one.
- Using simulated data with a known ground truth.
- Comparing across competing models (DiD, event study, synthetic control).

## Why This Matters
The goal isn’t to prove your model is the only correct one—it’s to show that your findings hold across reasonable assumptions. This strengthens credibility and provides stakeholders with the confidence to act on your results.

# Report Example

## Presenting Causal Results with Regression Discontinuity (RD) Design
When presenting causal effects without a true experiment, extra tests are needed to support findings:

- **Result**: A training program increased sales by about 19 units, a significant effect for those near the eligibility cutoff.

- **Method**: Used regression discontinuity—employees with at least 40 sales in Year 1 got the training; others did not.

- **Assumptions**: Local randomness around the cutoff and continuity of outcomes (except for the treatment effect).

- **Supportive Checks**:

    - Re-estimated results using different bandwidths (+10 and +5); estimates remained stable.

    - Checked that characteristics like part-time status were balanced just above/below the threshold.

    - Ran a placebo test at a fake cutoff (30 sales); found no significant effect, boosting confidence in the result.

- **Recommendation**: The training has a real impact for those near 40 sales. Consider expanding or better targeting the program based on these findings.

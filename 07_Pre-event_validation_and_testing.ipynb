# Testing Assumptions in Difference-in-Differences

## Parallel Trends Assumption
For Difference-in-Differences, our key assumption is the parallel trends assumption: that treated and untreated groups would have followed the same trajectory if no treatment had occurred.

We can’t observe this directly, but we can test for supporting evidence.

## Tools for Testing Assumptions

### Pre-Event Testing
- Use the pre-treatment period as a testing ground.
- Check whether the treated and untreated groups follow similar patterns before the intervention.



### Placebo Tests
- Assign fake treatment dates to test whether the model incorrectly detects effects.
- If the model finds significant impacts where no treatment occurred, that’s a red flag for bias or misspecification.



### Why This Matters
Testing our assumptions helps build confidence in causal claims and identify problems before drawing conclusions. It applies to all causal inference models, not just DiD.

## Testing the Parallel Trends Assumption

### Visual Inspection
- Plot outcomes over time for treated and untreated groups.
- If lines move together at similar rates, that's a good sign-but not proof of parallel trends.

### Important Caution
- "Not statistically significant" $\neq$ "trends are identical."
- Noisy pre-period data can mask real differences; use visuals and tests together.

### Code-Based Test (Pre-Period Only)
1. Subset to pre-treatment dates
2. Create trend (often centered so the event date = 0; earlier dates = -1, -2, ...)
3. Fit regression in the pre-period: `outcome ~ treated + trend + treated_trend`
4. Focus on the interaction coefficient (`treated_trend`): this estimates the difference in trends between groups before treatment

### Interpreting the Test
- Significant `treated_trend` → evidence against parallel trends (violation)
- Not significant → encouraging but not definitive; you didn’t detect a difference, which is not the same as proving equality

**Takeaway**: Combine visual inspection, pre-period regression tests, and robustness checks to evaluate parallel trends; no single diagnostic is sufficient on its own.

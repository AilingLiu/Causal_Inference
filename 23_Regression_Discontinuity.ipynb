# Regression Discontinuity
Unlike Event Studies or Synthetic Controls, Regression Discontinuity (RD) does not rely on time series data. Instead, it focuses on a running variable—a continuous measure with a threshold that determines who receives treatment.


For example:

- Students receive a scholarship if their test score ≥ 70.
- Customers get free shipping if they spend over $50.
- Sales reps receive advanced training if their past sales exceed a threshold.
- Golfers qualify for a tournament if their score is below a cutoff.

## Local Randomization
- In strict RD, treatment is deterministic: everyone above the cutoff gets treated, everyone below does not.
- Around the cutoff, however, randomness creeps in:
- Some students just above may have gotten lucky.
- Some just below may have had a bad day.
- These small random factors blur systematic differences, making individuals just above and below the cutoff comparable in expectation.
- We do not expect that individuals far above the cutoff are comparable to individuals far below the cutoff



## Why RD Works
- Near the threshold, treatment assignment is “as good as random.”
- This local randomization assumption underpins RD’s credibility, similar to how parallel trends supports Difference-in-Differences.

# Components of Regression Discontinuity (RD) Design
Regression discontinuity leverages cutoffs in a running variable to estimate causal effects.

## Running Variable and Cutoff
- Running variable: a continuous measure that determines treatment assignment (e.g., test score, income, app usage).
- Cutoff (threshold): treatment status changes sharply above or below a threshold.
- Example: students scoring ≥ 80 receive a scholarship; users making ≥ 10 purchases get a discount.



## Estimating the Treatment Effect
- Compare outcomes just above and just below the cutoff controlling for the running variable.
- Plotting outcome vs. running variable should show two smooth lines with a jump at the cutoff.
- That jump = the estimated causal effect.
- This effect is local: it applies only to units near the cutoff (Local Average Treatment Effect, LATE). We do not expect the jump would be of a similar size if we shifted the cutoff.



## Sharp vs. Fuzzy Designs
**Sharp RD**
- In sharp RD, treatment status changes deterministically (everyone above cutoff treated, everyone below untreated).

**Fuzzy RD**
- In fuzzy RD, the probability of treatment jumps at the cutoff but not perfectly (eligibility vs. actual uptake). For example, some students may be eligible for the scholarship but don't take it. Method: Two-Stage Least Squares (2SLS)

## Key Assumption
- Local randomization: units just above and just below the cutoff are comparable.
- Risk: manipulation around the cutoff (e.g., highly motivated individuals just below threshold finding ways to get treated).

## Resources
- **Local Average Treatment Effect (LATE)**: LATE is the treatment effect estimated for individuals whose treatment status is influenced by a cutoff or instrument, rather than the entire population. [Wikipedia: Local average treatment effect](https://en.wikipedia.org/wiki/Local_average_treatment_effect)
- **Two-Stage Least Squares (2SLS)**: 2SLS is an econometric method used to estimate causal effects when treatment is not perfectly assigned (e.g., fuzzy RD). In the first stage, treatment assignment is predicted using an instrument (like eligibility at the cutoff). In the second stage, the predicted treatment is used to estimate the outcome. # Estimating Treatment Effects with Regression Discontinuity

Regression Discontinuity (RD) estimates causal effects by comparing outcomes for units just above and just below a cutoff in a running variable (e.g., test scores, income). The sharp RD model assumes treatment assignment is fully determined by crossing this threshold.

## Model Setup
- Outcome ($Y_i$): the variable of interest (e.g., graduation, revenue).
- Running variable: determines treatment eligibility (e.g., score, income).
- Treated indicator: equals 1 if the unit is on the treated side of the cutoff, 0 otherwise.
- Bandwidth: restricts the data to a window around the cutoff (e.g., ±5 points) to satisfy the local randomization assumption.

## Linear Model
A typical RD regression is fit using OLS:

$Y_i = \beta_o + \beta_1 * RunningVar_i + \beta_2 * Treated_i + \epsilon_i$

- $\beta_o$: expected outcome just below the cutoff (when RunningVar = 0 after recentering).
- $\beta_1$: captures the relationship between the running variable and the outcome just below the cutoff.
- $\beta_2$: the size of the jump at the cutoff-our local average treatment effect (LATE).

## Interpretation
- A significant $\beta_2$ suggests a causal effect at the cutoff.
- The effect is local, valid only for individuals near the threshold.
- Bandwidth choice is critical:
    - Too narrow → not enough data.
    - Too wide → risks violating the local randomization assumption.
- This framework lets us isolate causal impacts at eligibility thresholds while accounting for pre-existing trends in the outcome.

## Recentering the Running Variable
- While not necessary, it is common to recenter the running variable so that it takes the value of zero at the cutoff
- This only changes the interpretation of the intercept—not the treatment effect

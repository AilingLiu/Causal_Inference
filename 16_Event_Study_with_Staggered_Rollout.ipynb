# Modeling Staggered Rollouts

## Event Studies with Staggered Rollouts
In many real-world cases, treatments roll out gradually across units (markets, users, or companies) rather than all at once. These are called staggered rollouts.

## Examples
- Self-driving ride-sharing launched city by city
- Firms making earnings announcements on different dates
- A product feature tested in limited markets before a global launch



## Key idea
- Re-align time so that all units share a common treatment point, time zero.
- Define relative time:
    - 0 = event date for each unit.
    - -1, -2, ... = periods before.
    - +1, +2, ... = periods after.



## Implementation
- Create relative pre/post indicators instead of calendar dates.
- Add dummy variables for windows (e.g., pre2, pre1, post0, post1, post2).
- Estimate the event study using these relative indicators.



## Practical Considerations
- Be consistent about time units (e.g., is the Monday following a Friday event date treated as +1 or +3?).
- Staggered rollout models are ideal when:
    - Multiple units are treated at different times,
    - The treatment is functionally the same across units.

## Benefits
- Allows borrowing strength across units.
- Preserves the pre-post causal logic while handling varied event dates.


# Sensitivity Analysis and Hyperparameter Choices in Causal Models

- When estimating causal models, analysts must make hyperparameter choices—decisions not part of the model itself but that affect results.

## Examples of Hyperparameters
- Length of the pre-treatment window (e.g., 14, 30, 60 days)
- Choice of time unit (e.g., daily, weekly, monthly data)
- Treatment of outliers or applying data transformations
- Length of the post-period window if no natural cutoff exists
- Selection of comparison groups

## Why Sensitivity Checks Matter
- Results may shift depending on these choices.
- Stability across reasonable variations increases confidence in findings.
- Sensitivity checks demonstrate that estimated effects are not artifacts of arbitrary decisions.

## Practical Steps
- Re-run models with different pre- and post-period windows.
- Test alternative time granularities (daily vs. weekly).
- Swap in different--but reasonable--control groups if available.
- Check robustness when excluding or transforming outliers.

## Communicating with Stakeholders
- Sensitivity analysis helps address skepticism. Robust results build credibility and show the effect is real, not fragile.

# Validating Event Study Models with Placebo Tests
Event studies must be validated to ensure that estimated effects are real and not artifacts of noise or unstable comparisons. Three key validation tools help strengthen confidence:

1. Pre-period indicators
- Add dummy variables for periods before the event (e.g., pre1, pre2).
- Expected coefficients: close to zero.
- Large or significant values suggest:
    - Anticipatory effects (e.g., leaked earnings influencing stock prices).
    - Model misspecification, similar to failing a parallel trends check.



2. Single placebo tests
- Remove post-treatment data.
- Select a fake event date in the pre-period and re-estimate the model.
- If strong effects appear where no treatment occurred → red flag for spurious correlations or over-sensitivity to noise.



3. Distribution of placebo effects
- Run many placebo tests by:
    - Randomizing event dates in the pre-period.
    - Randomly assigning treated units.
- Collect all placebo estimates into a distribution.
- Your real effect should stand out (e.g., in the top 5% of estimates).

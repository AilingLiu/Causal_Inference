# DAG & Potential Outcomes Framework

## Two Languages of Causality

In causal inference, there are two main ways to describe causal relationships:

- Directed acyclic graphs (DAGs) – visual tools common in computer science and machine learning
- Potential outcomes framework – counterfactual reasoning often used in economics and policy research

We’ll use both: DAGs to visualize relationships, and potential outcomes to define and interpret models.


## DAG

Key Concepts in DAGs
1. Confounders
- A confounder influences both the treatment and the outcome.
- Example: Brand affinity affects both whether someone joins a loyalty program and how much they spend.
- Failing to control for the confounder can lead to spurious effects.

2. Colliders
- A collider is a variable caused by two other variables (two arrows pointing in).
- Example: Brand sponsorships depend on both social media followers and goals scored.
- Conditioning on the collider (e.g., only analyzing sponsored athletes) can create bias that wasn’t there before.
    - Suppose both social media followers and goalsscored affect whether an athlete gets brand sponsorships. That's two arrows pointing into brand sponsorships. This makes brand sponsorships a collider. It's a variable caused by two other variables. If we condition on the collider, say, by analyzing only athletes who received sponsorships, it can create biased associations between goals scored and social media followers. Specifically, there's likely a positive relationship between social media popularity and goal scored. However, by conditioning on the collider, athletes with brand sponsorships, we might even find a negative relationship between popularity and goals. This is a classic mistake in causal inference. Conditioning on a collider can open up a biasing path that wasn't there before.

3. Mediators
- A mediator lies on the causal path between two variables.
- Example: Family wealth → SAT scores → College attendance.
- Researchers can estimate:
    - The total effect (direct + indirect through the mediator).
    - Or the direct effect (controlling for the mediator).

**Why DAGs Matter**
- Directed: arrows define the direction of cause and effect.
- Acyclic: arrows don’t loop back.
- DAGs clarify what to control for depending on whether we want total or direct effects.
- They help prevent common mistakes like confounding bias or collider bias.

In summary: Confounders cause both variables, mediators are part of the causal path, and colliders are caused by multiple upstream factors.

## Potential Outcomes Framework

**The Potential Outcomes Framework**

Causal inference can also be described through potential outcomes.

For each unit, there are two possibilities:
- Y(1): the outcome if treated
- Y(0): the outcome if untreated
- The causal effect is the difference Y(1) – Y(0).

The challenge: we only ever observe one of these for each person. This is the fundamental problem of causal inference.

**Counterfactuals and Estimation**
- If someone is treated, we observe Y(1) and never see Y(0).
- If untreated, we observe Y(0) and never see Y(1).
- The key task is estimating the missing counterfactual.
- This can be done by using comparison groups that approximate what would have happened in the unobserved scenario.

**DAGs vs. Potential Outcomes**
- DAGs: visualize relationships and guide decisions on what to control for (e.g., confounders vs. mediators vs. colliders).
- Potential outcomes: focus directly on counterfactuals, needing less structural detail but requiring good comparison groups.
    - The counterfactual is the potential outcome we don’t observe — what would have happened under a different treatment.
    - Can we find untreated groups or individuals who are similar enough to predict what would have happened to the treated ones?
- Together, these two “languages of causality” complement each other: one clarifies the structure, the other frames estimation.
